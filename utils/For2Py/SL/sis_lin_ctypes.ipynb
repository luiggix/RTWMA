```python
import ctypes
import numpy as np

# Cargar DLL
lib = ctypes.CDLL("./sistema_lineal.dll")

# Define argtypes sin flags estrictos en los arrays 2D
lib.crear_sistema.argtypes = [ctypes.POINTER(ctypes.c_void_p)]

lib.inicializar_sistema.argtypes = [
    ctypes.c_void_p,
    ctypes.c_int,
#    np.ctypeslib.ndpointer(dtype=np.float64, ndim=2, flags=['F_CONTIGUOUS', 'WRITEABLE']),  # sin C_CONTIGUOUS ni F_CONTIGUOUS
#    np.ctypeslib.ndpointer(dtype=np.float64, ndim=1, flags=['C_CONTIGUOUS', 'WRITEABLE'])
]

lib.resolver_jacobi.argtypes = [
    ctypes.c_void_p,
    np.ctypeslib.ndpointer(dtype=np.float64, ndim=1, flags=['C_CONTIGUOUS', 'WRITEABLE']),
    ctypes.c_int,
    ctypes.c_double
]

lib.destruir_sistema.argtypes = [ctypes.c_void_p]

# Crear sistema
sistema_ptr = ctypes.c_void_p()
print(sistema_ptr)
lib.crear_sistema(ctypes.byref(sistema_ptr))
print(sistema_ptr)


# Define matrices y vectores con orden Fortran para A, sin flags estrictos
A = np.array([[10.0, -1.0, 2.0],
              [-1.0, 11.0, -1.0],
              [2.0, -1.0, 10.0]], dtype=np.float64, order='F')

b = np.array([6.0, 25.0, -11.0], dtype=np.float64, order='C')

x_out = np.zeros(3, dtype=np.float64, order='C')

print("A flags:", A.flags)  # asegúrate que diga F_CONTIGUOUS
print(b.flags)

# Llamadas
#lib.inicializar_sistema(sistema_ptr, 3, A, b)
lib.inicializar_sistema(sistema_ptr, 3)
lib.resolver_jacobi(sistema_ptr, x_out, 100, 1e-6)

print("Solución:", x_out)

lib.destruir_sistema(sistema_ptr)

```
