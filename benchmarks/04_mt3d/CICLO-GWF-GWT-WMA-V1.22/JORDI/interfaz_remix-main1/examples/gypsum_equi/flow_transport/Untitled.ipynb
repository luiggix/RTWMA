```python
# ====================================
# LECTURA DE ARCHIVO .m3d CON VARIAS ESPECIES (MT3D-USGS)
# ====================================

import os
import numpy as np
from flopy.utils import Mt3dMsUcnFile

# Ruta al archivo .m3d
archivo_m3d = "transport.m3d"
ucn = Mt3dMsUcnFile(archivo_m3d)

# Obtener tiempos y especies
tiempos = ucn.get_times()
nspecies = ucn.nspecies

print(f"⏱ Tiempos disponibles: {tiempos}")
print(f"🧪 Número de especies: {nspecies}")

# ===========
# ESCRITURA
# ===========

archivo_salida = "transport_dos_especies.txt"
with open(archivo_salida, "w") as f:
    for t in tiempos:
        for comp in range(nspecies):
            nombre = "Sulfato" if comp == 0 else "Calcio"
            f.write(f"FOR COMPONENT NO. {comp+1:02d} – {nombre}\n")
            f.write(f"TOTAL ELAPSED TIME SINCE BEGINNING OF SIMULATION = {t:.6f} D\n")
            f.write("LAYER 1:\n")

            arr = ucn.get_data(totim=t, mcomp=comp)
            valores = arr[0, 0, :]  # capa 0, fila 0, todas las columnas
            linea = " ".join([f"{v:.6g}" for v in valores])
            f.write(f"{linea}\n\n")

print(f"✅ Archivo generado: {archivo_salida}")
```
