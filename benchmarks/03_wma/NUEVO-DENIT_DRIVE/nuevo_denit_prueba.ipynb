## Generador del HTML

```python
import re

def generar_html(texto):

    secciones = re.split(r'\n# ', texto)

    html_secciones = []

    for i, sec in enumerate(secciones):

        if i == 0:
            titulo_sec = "Resumen"
            contenido = sec
        else:
            partes = sec.split("\n",1)
            titulo_sec = partes[0]
            contenido = partes[1] if len(partes)>1 else ""

        contenido = contenido.strip()

        # párrafos reales
        contenido = contenido.replace("\n\n","</p><p>")
        contenido = "<p>"+contenido+"</p>"

        # saltos de línea simples
        contenido = contenido.replace("\n","<br>")

        html_secciones.append(f"""
<div class="item-indice" data-sec="sec{i}">
{i+1}. {titulo_sec}
</div>

<div id="sec{i}" class="contenido">
{contenido}
</div>
""")

    html = f"""
<!DOCTYPE html>
<html lang="es">

<head>

<meta charset="UTF-8">

<title>{titulo.value}</title>

<script src="https://cdn.jsdelivr.net/npm/mathjax@3/es5/tex-mml-chtml.js"></script>

<script src="https://cdn.jsdelivr.net/npm/mathjax@3/es5/tex-mml-chtml.js"></script>
<style>

body {{
font-family: Arial;
margin: 30px;
}}

h1 {{
color:#17397f;
}}

.item-indice {{
font-size:20px;
margin:10px 0;
cursor:pointer;
color:#0044cc;
font-weight:bold;
}}

.contenido {{
display:none;
margin:15px 0 25px 20px;
padding:15px;
border-left:4px solid #0044cc;
background:#f5f7ff;
line-height:1.6;
}}

</style>

</head>

<body>

<h1>{titulo.value}</h1>

<hr>

{''.join(html_secciones)}

<script>

document.addEventListener("DOMContentLoaded", function() {{

document.querySelectorAll(".item-indice").forEach(function(el) {{

var id = el.dataset.sec;

el.addEventListener("click", function() {{
document.getElementById(id).style.display = "block";
}});

el.addEventListener("dblclick", function() {{
document.getElementById(id).style.display = "none";
}});

}});

document.querySelectorAll(".contenido").forEach(function(el) {{

el.addEventListener("dblclick", function() {{
this.style.display = "none";
}});

}});

}});

</script>

</body>
</html>
"""

    return html
```

## Botón para generar el HTML

```python
from pathlib import Path

def guardar_html(b):

    html = generar_html(editor.value)

    archivo = Path(f"{nombre}.html")

    archivo.write_text(html, encoding="utf8")

    print("HTML generado en:")
    print(archivo.resolve())

# Ejecutar dando clic en botón (guardar_html)
#boton = widgets.Button(description="Generar HTML")
#display(boton)
#boton.on_click(guardar_html)

# Ejecutar automáticamente
guardar_html(None)
```
