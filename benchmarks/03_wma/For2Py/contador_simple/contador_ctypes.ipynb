```python
import ctypes

# Cargar la biblioteca
lib = ctypes.CDLL("./libcontador.dll")

# Definir tipos
lib.crear_contador.argtypes = [ctypes.POINTER(ctypes.c_void_p)]
lib.destruir_contador.argtypes = [ctypes.c_void_p]
lib.incrementar_c.argtypes = [ctypes.c_void_p]
lib.obtener_valor.argtypes = [ctypes.c_void_p, ctypes.POINTER(ctypes.c_int)]

# Crear instancia
obj = ctypes.c_void_p()
lib.crear_contador(ctypes.byref(obj))

# Incrementar y mostrar valor
for i in range(5):
    lib.incrementar_c(obj)
    valor = ctypes.c_int()
    lib.obtener_valor(obj, ctypes.byref(valor))
    print(f"Iteración {i+1}: valor = {valor.value}")

# Liberar memoria
lib.destruir_contador(obj)
```
