```python
from cffi import FFI

ffi = FFI()

# Declarar las funciones y tipos que vamos a usar (la interfaz C)
ffi.cdef("""
    void crear_contador(void **ptr);
    void destruir_contador(void *ptr);
    void incrementar_c(void *ptr);
    void obtener_valor(void *ptr, int *valor);
""")

# Cargar la biblioteca compilada
lib = ffi.dlopen("libcontador.dll")

# Crear el objeto (puntero opaco en C)
contador_ptr = ffi.new("void **")
lib.crear_contador(contador_ptr)
contador = contador_ptr[0]

# Usar el objeto
for i in range(5):
    lib.incrementar_c(contador)
    valor = ffi.new("int *")
    lib.obtener_valor(contador, valor)
    print(f"Iteración {i+1}: valor = {valor[0]}")

# Liberar el objeto
lib.destruir_contador(contador)
```
