```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
denit_excel = pd.read_excel("DENIT-10.xlsx", skiprows=2, header=None).drop(columns=[0])
spe_names = [c.strip() for c in denit_excel.iloc[0:10,0]]
times = ["0.00d", "0.01d", "0.05d", "0.10d", "1.00d"]
ranges = [(0, 11), (21,32), (46,57), (71,82), (96,107)]
rdict = dict(zip(times, ranges)) 
nspec = 10
ncomp = 7
```

```python
for it in range(0,5):
    isp = ranges[it][0]
    icp = ranges[it][1]
    s_t0 = denit_excel.iloc[isp:isp+nspec, 1:11].to_numpy(dtype=float)
    c_t0 = denit_excel.iloc[icp:icp+ncomp, 1:11].to_numpy(dtype=float)
    x = np.linspace(0, 1.0, 10)
    
    for i, c in enumerate(c_t0):
        plt.plot(x, c, marker=".", lw = (i+1)*0.75, label = f"C{i+1}", zorder = 10-i)
    plt.title(f"Components. Time = {times[it]}")
    plt.grid()
    plt.yscale("log")
    plt.legend()
    plt.show()
```

```python
for i, sn in enumerate(spe_names):
    if sn != 'h2o' and sn != 'ca+2':
        plt.plot(x, s_t0[i], marker="o", label = sn)
plt.title("Species")
plt.grid()
plt.legend()
plt.yscale("log")
plt.show()
```
