```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#
# Cargar los datos desde archivos
# -----
c1 = np.load('c1_exact_200.npy')
c2 = np.load('c2_exact_200.npy')

#
# Dimensiones de los datos
# -----
steps = c1.shape[0]
Nx = c2.shape[1]

#
# Dimensiones espacial y temporal
# -----
x = np.linspace(0,30,Nx)
#x = np.array([i - 0.5 for i in range(1, Nx + 1)], dtype=float)
times = np.array([i for i in range(0, steps)])
```

```python
#
# Definir la figura
# -----
fig, ax = plt.subplots(2, 1, sharex=True, figsize=(7,6))

#
# Graficar todos los pasos
# -----
color = "silver"
lw = 0.75
for t in times:
    ax[0].plot(x, c1[t], c = color, lw = lw)
    ax[1].plot(x, c2[t], c = color, lw = lw)

#
# Graficar el primer paso
# -----
tsel = 0
color = "royalblue"
lw = 1.5
ax[0].plot(x, c1[tsel], "o-", c = color, mec="black", mfc=color, markersize=5,
           lw = lw, label = f"$t$ = {tsel}")
ax[1].plot(x, c2[tsel], "o-", c = color, mec="black", mfc=color, markersize=5,
           lw = lw, label = f"$t$ = {tsel}")

#
# Graficar un paso específico
# -----
tsel = 10
color = "orangered"
lw = 1.5
ax[0].plot(x, c1[tsel], "s-", c = color, mec="black", mfc=color, markersize=5,
           lw = lw, label = f"$t$ = {tsel}")
ax[1].plot(x, c2[tsel], "s-", c = color, mec="black", mfc=color, markersize=5,
           lw = lw, label = f"$t$ = {tsel}")

#
# Decoración de la gráfica
# -----
ax[0].grid()
ax[1].grid()

ax[0].legend(loc="upper left")
ax[1].legend()

ax[0].spines.right.set_visible(False)
ax[0].spines.top.set_visible(False)

ax[1].spines.right.set_visible(False)
ax[1].spines.top.set_visible(False)

ax[0].set_ylabel("$Ca^{2+}$") 
ax[1].set_ylabel("$SO_4^{2-}$")

plt.suptitle("Exact solution", x=0.085, ha="left")
plt.tight_layout()
plt.savefig("exact_sol_gypsum.pdf")
plt.show()
```
