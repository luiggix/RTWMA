Solución analítica.

$$
u(x,t)=u_0+\frac{M(t)}{\sqrt{4\pi D_L t}}\exp\!\left(-\frac{(x-x_0-Vt)^2}{4D_L t}\right), \hspace{10mm} M(t)=\frac{u_e\,V_e}{4n\pi D_T\,t} = M0
$$

Especies.

$$c_1 = \frac{u+\sqrt{u^2 + 4K}}{2}, \hspace{5mm} c_1 = \frac{-u+\sqrt{u^2 + 4K}}{2} $$


```python
#
# Dimensiones del problema
# -----
Nx = 30
x = np.array([i - 0.5 for i in range(1, Nx + 1)], dtype=float)
times = [i for i in range(0, 41)]

# 
# Parámetros físicos
# -----
ue = 1.0
u0 = 1.0
V  = 0.4
Ve = 0.4
DL = 0.4
DT = 0.2
n  = 0.5
x0 = 10.5          # centro del pulso
M0 = 199.0         # "masa" inicial discreta en t=0 (impulso)
K = 1 / 30359.9

# 
# Cálculo de la solución
# -----
def u_sol(x, t):
    # --- Condición inicial discreta en t=0 ---
    if t == 0:
        u = np.full_like(x, u0, dtype=float)
        idx = int(np.argmin(np.abs(x - x0)))
        u[idx] = u0 + M0
        return u

    # --- Para t > 0 ---
    M_t = (ue * Ve ) / (4 * n * np.pi * DT * t)  # M(t)
    pref = M0 / np.sqrt(4 * np.pi * DL * t)
    expo = np.exp(-((x - x0 - V*t)**2) / (4 * DL * t))

    return u0 + pref * expo
```

```python
#
# Cálculo de la solución
# -----
u_pulso = {t: u_sol(x, t) for t in times}
c1_fun = lambda u, K: (u + np.sqrt(u*u + 4 * K) ) / 2
c2_fun = lambda u, K: (-u + np.sqrt(u*u + 4 * K) ) / 2

#
# Almacenamiento de la solución en archivos binarios
# -----
c1 = np.empty((len(times), len(x)))
c2 = np.empty((len(times), len(x)))

for t in times:
    c1[t] = c1_fun(u_pulso[t], K)
    c2[t] = c2_fun(u_pulso[t], K)

np.save('c1_exact.npy', c1)
np.save('c2_exact.npy', c2)
```
