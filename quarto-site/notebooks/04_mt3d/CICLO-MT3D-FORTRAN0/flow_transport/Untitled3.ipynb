<h1 style='font-size:26px;'>MODELO DE FLUJO Y TRANSPORTE USANDO MF6 Y MT3D-USGS PARA LAS ESPECIES SULFATO Y CALCIO</h1>


<details><summary style='font-size:20px;'>1️⃣  <a href='#Importación-de-librerías'>Importación de librerías</a></summary>
<p style='font-size:20px;'>Se importan las bibliotecas necesarias como `flopy`, `numpy`, `matplotlib`, `os` y utilerías de archivos binarios, que permiten construir, ejecutar y visualizar modelos de flujo y transporte subterráneo.</p>
</details>

<details><summary style='font-size:20px;'>2️⃣  <a href='#Definición-de-rutas-y-ejecutables'>Definición de rutas y ejecutables</a></summary>
<p style='font-size:20px;'>Se configuran las rutas y ubicación de los ejecutables MODFLOW 6 (mf6) y MT3D-USGS (mt3d-usgs_1.1.0_64.exe).</p>
</details>

<details><summary style='font-size:20px;'>3️⃣  <a href='#Configuración-de-la-simulación'>Configuración de la simulación</a></summary>
<p style='font-size:20px;'>Se define una simulación de flujo 1D usando MODFLOW 6 con: LX=30 m, LY=1 m, LZ=1 m, para un periodo de 40 días dividido en 40 pasos de tiempo.</p>
</details>

<details><summary style='font-size:20px;'>4️⃣  <a href='#Malla-y-propiedades-hidráulicas'>Malla y propiedades hidráulicas</a></summary>
<p style='font-size:20px;'>Se define la malla espacial y las propiedades hidráulicas del acuífero de 1 capa, 1 renglón y 30 columnas con una conductividad hidráulica = 1.0m/día y porosidad = 0.5.</p>
</details>

<details><summary style='font-size:20px;'>5️⃣  <a href='#Condiciones-de-frontera-y-pozos'>Condiciones de frontera</a></summary>
<p style='font-size:20px;'>Se especifican condiciones de frontera tipo Neumann de Flujo prescrito en la frontera izquierda CFI = 0.2 m³/d, inyectando un fluido con concentración específica para cada especie. En la frontera derecha se prescribe una carga hidráulica CFD = 1 m.</p>
</details>

<h2 style='font-size:22px;'> Modelo de Flujo MODFLOW 6</h2>

<details><summary style='font-size:20px;'>6️⃣  <a href='#Ejecución-del-modelo-de-flujo'>Ejecución del modelo de flujo</a></summary>
<p style='font-size:20px;'>Se escriben los archivos de entrada del modelo y se ejecuta MODFLOW 6. Esto permite resolver la ecuación de flujo y calcular la distribución de cargas hidráulicas.</p>
</details>

<details><summary style='font-size:20px;'>7️⃣  <a href='#Visualización-de-resultados-hidráulicos'>Visualización de resultados hidráulicos</a></summary>
<p style='font-size:20px;'>Se visualizan los resultados de carga hidráulica y descarga específica obtenidos de los archivos `.hds` y `.bud`.</p>
</details>

<h2 style='font-size:22px;'> Modelo de Transporte MT3D-USGS</h2>

<details><summary style='font-size:20px;'>8️⃣  <a href='#Configuración-del-modelo-de-transporte'>Configuración del modelo de transporte</a></summary>
<p style='font-size:20px;'>Se crea el modelo de transporte asociado al flujo, incluyendo las especies sulfato y calcio. Esto permite simular cómo las concentraciones evolucionan en el tiempo dentro del flujo de agua subterránea.</p>
</details>

<details><summary style='font-size:20px;'>9️⃣  <a href='#Condiciones-iniciales-y-de-frontera'>Condiciones iniciales y de frontera</a></summary>
<p style='font-size:20px;'>Se establecen condiciones iniciales y de frontera para cada especie: c₁ para el sulfato y c₂ para el calcio, y condiciones iniciales y de frontera de concentración prescrita c₁ = 1.0000329 y c₂ = 3.294 × 10⁻⁵, con pulsos en la celda 11 con c₁₁₂ᵢ = 200 para el sulfato y c₂₁₂ᵢ = 1.647 × 10⁻⁷ para el calcio.</p>
</details>

<details><summary style='font-size:20px;'>🔟  <a href='#Advección,-dispersión-y-parámetros'>Advección, dispersión y parámetros</a></summary>
<p style='font-size:20px;'>Se definen los esquemas de advección y dispersión, junto con parámetros como dispersividad y difusión molecular. Esto determina cómo se mueve y se mezcla el contaminante con el agua en el acuífero.</p>
</details>

<h2 style='font-size:22px;'> Reacciones Químicas</h2>

<details><summary style='font-size:20px;'>1️⃣1️⃣  <a href='#Ejecución-y-resultados-sin-reacciones'>Ejecución y resultados sin reacciones</a></summary>
<p style='font-size:20px;'>Se ejecuta MT3D-USGS sin el paquete de reacciones (RCT) para obtener un caso base de transporte puro, sin transformación química. Sirve como referencia para evaluar el impacto de las reacciones.</p>
</details>

<details><summary style='font-size:20px;'>1️⃣2️⃣  <a href='#Inclusión-del-paquete-RCT'>Inclusión del paquete RCT</a></summary>
<p style='font-size:20px;'>Se activa el paquete RCT para incluir adsorción lineal y decaimiento de primer orden. Esto permite simular la interacción química entre el contaminante y el medio poroso.</p>
</details>

<details><summary style='font-size:20px;'>1️⃣3️⃣  <a href='#Configuración-de-adsorción-y-decaimiento'>Configuración de adsorción y decaimiento</a></summary>
<p style='font-size:20px;'>Se asignan parámetros como la densidad aparente del suelo (`rhob`) y coeficientes de reacción (`rc1`, `rc2`) para modelar los procesos químicos de cada especie. Esto afecta la velocidad de retención o transformación del contaminante.</p>
</details>

<details><summary style='font-size:20px;'>1️⃣4️⃣  <a href='#Ejecución-del-modelo-con-reacciones'>Ejecución del modelo con reacciones</a></summary>
<p style='font-size:20px;'>Se ejecuta el modelo con reacciones habilitadas y se generan archivos `.m3d` con los resultados. Esto permite comparar los efectos de las reacciones frente al transporte puro.</p>
</details>

<p style='font-size:20px;'>Se grafican las concentraciones de las especies para el caso sin reacciones. Esto permite entender el comportamiento del transporte sin procesos químicos involucrados.</p>
</details>

<details><summary style='font-size:20px;'>1️⃣6️⃣  <a href='#Gráficas-con-reacciones'>Gráficas con reacciones</a></summary>
<p style='font-size:20px;'>Se visualizan las concentraciones simuladas con el paquete RCT, comparando su evolución con el caso sin reacciones. Esto ayuda a cuantificar el efecto de la adsorción y el decaimiento.</p>
</details>
